## Environments

Installing packages globally eventually breaks dependencies due to version conflicts between projects requiring different packages, as each package also requires others, until they are incompatible. We can use Miniforge to isolate project libraries into dedicated containers called **environments**.

- `mamba create -n env_name python=3.10` creates an isolated environment callen env_name with python 3.10 installed
- `mamba activate env_name` "launches" it
- `mamba install package_name` installs package called package_name from the conda-forge database
- `mamba deactivate` deactivates the environment, returning to base
- `mamba list` lists all packages in the current environment
- `mamba env list` lists all available environments
- `mamba env remove -n env_name` removes the environment specified

For complete documentation, more commands and troubleshooting, see this [link](https://mamba.readthedocs.io/en/latest/user_guide/mamba.html).

*You can replace `mamba` with `conda` if you're used to it or prefer it. The syntax is mostly the same.*

When a `.py` file is imported, Python immediately executes all code within it. To write safe modules, place executable pipeline commands inside an `if __name__ == '__main__':` block. This ensures the code only triggers if the script is run directly from the terminal, and prevents running code during imports.
